```python
lst = ["a", "b", "c", "d", "e"  ]

lst.append(5) #will add 5 to the end of the list
lst.insert(2, "new") #will insert "new" at index 2
lst.pop() #will remove and return the last item in the list
lst.extend([6, 7, 8]) #will add multiple items to the end of the list
print(lst)



#practice modifying number list
numbers = [10, 20, 30, 40]
numbers.append(50) #will add 50 to the end of the list
numbers.remove(20) #will remove the first occurrence of 20 from the list
print(numbers)

#practice slicing
fruits = ['apple', 'banana', 'cherry']
fruits[1:3] #will return ['banana', 'cherry'] index 1 to 2
fruits[:2] #will return ['apple', 'banana'] index 0 to 1
```

```python
words = ["cat", "dog", "elephant"]
result = [word.upper() for word in words if len(word)>3]
print(result) #will return ['ELEPHANT'] because only "elephant" has more than 3 characters 
#and it will be converted to uppercase
```

```python
#what is a dictionary?

x = {
    "name": "David", 
    "age": 10
    }
#"name" is the key and "David" is the value in the dictionary x
#what you define in the dictionary should be in key:value pairs, separated by commas
#dictionaries are like a actual dictionary so words (keys) and their meanings (values)
```

```python
#nested dictionary

x = {
    "Yaya": {"math": 100, "science": 95},
    "Momo": {"math": 95, "science": 100}
}
#defining a nested dictionary where the values of the main dictionary are also dictionaries

x["Yaya"] #will return {'math': 100, 'science': 95} which is the dictionary associated with the key "Yaya"
x["Yaya"]["math"] #will return 100 which is the value associated with the key "math" in the dictionary associated with "Yaya"
print(x["Yaya"]["math"]) #will print 100

#updating a dictionary

x["Yaya"]["science"]= 100
print(x) #will update the value of "science" for "Yaya" to 100 and print the updated dictionary
print(x["Yaya"]["science"]) #will print 100 which is the updated value for "science" for "Yaya"


#add a person to the dictionary
x["Alex"]= {"math": 90, "science": 85}
print(x) #will add a new key "Alex" with the associated dictionary {"math": 90, "science": 85} to the main dictionary x and print the updated dictionary


#print(x.keys()) #will print the keys of the main dictionary x which are "Yaya", "Momo", and "Alex"

#print(x.values()) #will print the values of the main dictionary x which are the dictionaries associated with "Yaya", "Momo", and "Alex"

#print(x.items()) #will print the key-value pairs of the main dictionary x 

#print(x.clear()) #will remove all items from the main dictionary x and leave it empty

#print(x.get("Yaya")) #will return the value associated with the key "Yaya" which is {'math': 100, 'science': 100}
#OR print(x["Yaya"]) #will also return the value associated with the key "Yaya" which is {'math': 100, 'science': 100}
#the above method will raise a KeyError if the key does not exist in the dictionary, while the get method will return None instead of raising an error

#x.pop("Alex") #will remove the key "Alex" and its associated value from the main dictionary x and return the removed value which is {'math': 90, 'science': 85}
```
