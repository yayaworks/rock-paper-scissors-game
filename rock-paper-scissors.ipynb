```python
import random

choices = ["rock", "paper", "scissors"]

user_action = input("Enter 'rock', 'paper', or 'scissors': " ).lower()

if user_action not in choices:
    print("Error: Invalid choice. Please choose rock, paper, or scissors.")
else:
    computer_action = random.choice(choices)
    print(f"You chose: {user_action}")
    print(f"Computer chose: {computer_action}")

    if user_action == computer_action:
        print("It's a tie!")
    elif (user_action ==  "rock" and computer_action == "scissors") or \
         (user_action == "paper" and computer_action == "rock") or \
         (user_action == "scissors" and computer_action == "paper"):
        print("You win!")
    else:
        print("You lose!")
```
